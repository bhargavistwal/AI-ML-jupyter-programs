## 7.Filtering Data

## 9. Mean,Max,Min

```python
import pandas as pd

data = {
    "Name": ["Aman","Riya","Rahul"],
    "Age": [19,20,18],
    "Marks": [88,92,75]
}

df = pd.DataFrame(data)
print("Mean Marks :", df["Marks"].mean())
print("Max Marks :", df["Marks"].max())
print("Min Marks :", df["Marks"].min())
```

## 10. Sorting

```python
import pandas as pd

data = {
    "Name": ["Aman","Riya","Rahul"],
    "Age": [19,20,18],
    "Marks": [88,92,75]
}

df = pd.DataFrame(data)
df_sorted = df.sort_values(by="Marks", ascending=False)
print(df_sorted)
```

## 11 Handling Missing Values

```python
import pandas as pd

# Create DataFrame
data = {
    "Name": ["Aman", "Riya", "Rahul", "Sneha"],
    "Marks": [85, 90, 75, 88]
}

df = pd.DataFrame(data)

print("Original DataFrame:\n")
print(df)


# 1. Add Missing Value
df.loc[2, "Marks"] = None   # Add missing value at index 2

print("\nAfter Adding Missing Value:\n")
print(df)


# 2. Detect Missing Values
print("\nMissing Value Count:\n")
print(df.isnull().sum())


# 3. Replace Missing Value (Fill with Mean)
df["Marks"] = df["Marks"].fillna(df["Marks"].mean())

print("\nAfter Replacing Missing Value with Mean:\n")
print(df)


# 4. Again Add Missing Value (for drop example)
df.loc[1, "Marks"] = None

print("\nAfter Adding Another Missing Value:\n")
print(df)


# 5. Drop Rows with Missing Value
df_dropped = df.dropna()

print("\nAfter Dropping Missing Value:\n")
print(df_dropped)
```
